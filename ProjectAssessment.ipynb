```python
import seaborn as sns
```

```python
movies = pd.read_csv('Movie+Assignment+Data.csv')
```

```python
print(movies.head())
```

```python
movies['Profit'] = (movies['Gross'] - movies['budget']) / 1_000_000
```

```python
movies['ROI'] = (movies['Profit'] * 1_000_000 / movies['budget']) * 100
```

```python
top_10_profit = movies.sort_values(by='Profit', ascending=False).head(10)
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(data=top_10_profit, x='Profit', y='Title', palette='viridis')
plt.title('Top 10 Most Profitable Movies ($ Millions)')
plt.show()
```

```python
genre_perf = movies.groupby('genre_1')[['Profit', 'IMDb_rating']].mean().sort_values(by='Profit', ascending=False)
print(genre_perf)
```

```python
genre_perf['Profit'].plot(kind='bar', color='orange', title='Average Profit by Genre')
plt.ylabel('Profit in Millions')
plt.show()
```

```python
demo_votes = movies[['VotesU18', 'Votes1829', 'Votes3044', 'Votes45A']].mean()
demo_votes.plot(kind='line', marker='o', title='Average Rating by Age Group')
plt.show()
```
